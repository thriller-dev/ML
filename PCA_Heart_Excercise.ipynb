```python
import numpy as np
import pandas as pd
df = pd.read_csv('/content/heart.csv')
df.head()
```

```python
df.columns
```

```python
from scipy.stats import zscore
numerical_cols = ['Age','RestingBP','Cholesterol','FastingBS','MaxHR','Oldpeak']
z_scores = np.abs(df[numerical_cols].apply(zscore))
df_filtered = df[(z_scores < 3).all(axis=1)]
print(f"Original dataset size: {df.shape[0]} rows")
print(f"Filtered dataset size: {df_filtered.shape[0]} rows")
df_filtered.to_csv("cleaned_data.csv", index=False)
```

```python
df_filtered.head()
```

```python
df_filtered['ChestPainType'].unique()
df_filtered['ST_Slope'].unique()
```

```python
df_filtered['RestingECG'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder
df1 = df_filtered.copy()
label_encoder = LabelEncoder()
df1['Sex'] = label_encoder.fit_transform(df1['Sex'])
df1['ExerciseAngina'] = label_encoder.fit_transform(df1['ExerciseAngina'])
df1.head()
```

```python
chest_pain_mapping = {'TA': 0, 'ATA': 1, 'NAP': 2, 'ASY': 3}
resting_ecg_mapping = {'Normal': 0, 'ST': 1, 'LVH': 2}
st_slope_mapping = {'Up': 0, 'Flat': 1, 'Down': 2}
df1['ChestPainType'] = df['ChestPainType'].map(chest_pain_mapping)
df1['RestingECG'] = df['RestingECG'].map(resting_ecg_mapping)
df1['ST_Slope'] = df['ST_Slope'].map(st_slope_mapping)
```

```python
df1.head()
```

```python
print(type(df1))
```

```python
df1.head()
```

```python
X = df1.drop("HeartDisease",axis='columns')
y = df1.HeartDisease

X.head()
```

```python
y.head()
```

```python
from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
X_scaled = scalar.fit_transform(X)
X_scaled
```

```python
print(type(X_scaled))
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=30)
```

```python
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
svm_model = SVC(kernel='linear', random_state=42)
svm_model.fit(X_train, y_train)
svm_model.score(X_test,y_test)
```

```python
rf_model = RandomForestClassifier()
rf_model.fit(X_train, y_train)
rf_model.score(X_test,y_test)
```

```python
lr_model = LogisticRegression()
lr_model.fit(X_train, y_train)
lr_model.score(X_test,y_test)
```

```python
from sklearn.decomposition import PCA
pca = PCA(0.95)
X_pca = pca.fit_transform(X)
```

```python
X_train_pca, X_test_pca, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=30)
```

```python
svm_model = SVC(kernel='linear', random_state=42)
svm_model.fit(X_train_pca, y_train)
svm_model.score(X_test_pca,y_test)
```

```python
rf_model = RandomForestClassifier()
rf_model.fit(X_train_pca, y_train)
rf_model.score(X_test_pca,y_test)
```

```python
lr_model = LogisticRegression()
lr_model.fit(X_train_pca, y_train)
lr_model.score(X_test_pca,y_test)
```
